## Decision Tree Classifier Implementation

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
iris
```

```python
print(iris['DESCR'])
```

```python
print(iris['target'])
```

```python
# independent features
## convert the data,array into DF and give column name
X=pd.DataFrame(iris['data'],columns=['sepal length','sepal width','petal length','petal width'])
```

```python
# independent feature
y=iris['target']
```

```python
# train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=10)
```

```python
# In decision tree, standaralization and normalization is not required. Directly apply the Decision Tree algorithm
```

```python
# apply Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier
treeclassifier=DecisionTreeClassifier()
```

```python
treeclassifier.fit(X_train,y_train)
```

```python
# Visualize the Decision Tree

from sklearn import tree
plt.figure(figsize=(15,20))
tree.plot_tree(treeclassifier,filled=True)
```

```python
# gini value lies between 0 to 0.5 and entropy lies b/w 0 to 1
# If gini=0, means split is pure split
```

```python
# Max_depth=2

# apply Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier
treeclassifier2=DecisionTreeClassifier(max_depth=2)
```

```python
treeclassifier2.fit(X_train,y_train)
```

```python
# Visualize the Decision Tree

from sklearn import tree
plt.figure(figsize=(15,20))
tree.plot_tree(treeclassifier2,filled=True)
```

```python
y_pred=treeclassifier2.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```
